

For this assignment, I chose to use Python within Google Colab to analyze the mock healthcare dataset because it provides an efficient and structured way to organize and interpret data. The pandas library allows healthcare data to be stored in a DataFrame, which closely resembles real-world clinical datasets and makes the information easy to read, label, and manage. Each variable, such as age and heart rate, can be accessed directly, reducing confusion and improving clarity during analysis.

Python also ensured accuracy and transparency in the calculations. To obtain the results, the dataset was first loaded into a DataFrame, allowing numerical columns to be isolated for analysis. The average age and average heart rate were calculated by summing all values in their respective columns and dividing by the total number of patients. Python’s built-in functions, such as .sum(), .mean(), and len(), were used to display both the intermediate values and the final results. This approach clearly shows how the totals and averages were derived rather than only presenting the final numbers.

Displaying the sums and patient count helped verify that the calculations were correct and allowed the results to be easily explained and replicated. This step-by-step process improves transparency and aligns with best practices in healthcare data analysis, where accuracy and traceability are essential.

Another advantage of using Python is scalability. Although this dataset contained only five patients, the same code could be applied to much larger datasets without changes. Overall, Python provided a reliable, organized, and reproducible method for obtaining accurate results from healthcare data.


```python
# Import the pandas library for data handling and analysis
import pandas as pd

# Create a dictionary containing the mock healthcare data
data = {
    "PatientID": [1001, 1002, 1003, 1004, 1005],
    "Age": [45, 62, 28, 55, 39],
    "Gender": ["Female", "Male", "Female", "Male", "Female"],
    "ZipCode": [12345, 23456, 34567, 45678, 56789],
    "VisitDate": ["2024-01-15", "2024-01-15", "2024-01-16", "2024-01-16", "2024-01-17"],
    "VisitType": ["Regular Checkup", "Follow-up", "Urgent Care", "Regular Checkup", "Follow-up"],
    "PrimaryConcern": [
        "Annual Physical",
        "Diabetes Management",
        "Flu Symptoms",
        "Preventive Care",
        "Hypertension Check"
    ],
    "BloodPressureSystolic": [120, 138, 118, 142, 128],
    "BloodPressureDiastolic": [80, 88, 76, 92, 84],
    "HeartRate": [72, 78, 88, 76, 70],
    "BMI": [24.5, 29.2, 22.1, 31.5, 26.8],
    "Smoker": ["No", "Former", "No", "Yes", "No"],
    "ChronicCondition": ["No", "Yes", "No", "Yes", "Yes"],
    "MedicationsPrescribed": [1, 3, 2, 2, 1],
    "FollowUpRecommended": ["No", "Yes", "Yes", "Yes", "Yes"]
}

# Convert the dictionary into a pandas DataFrame
df = pd.DataFrame(data)

# Display the dataset to verify it loaded correctly
df
```

```python
# Calculate the average age of patients
average_age = df["Age"].mean()

# Calculate the average heart rate of patients
average_heart_rate = df["HeartRate"].mean()

# Display the calculated averages
average_age, average_heart_rate
```

```python
# Calculate the total sum of all ages
age_sum = df["Age"].sum()

# Calculate the total sum of all heart rate values
heart_rate_sum = df["HeartRate"].sum()

# Count the total number of patients in the dataset
patient_count = len(df)

# Display the intermediate values used in the calculations
age_sum, heart_rate_sum, patient_count
```

```python
# Print the final results with clear explanations
print(f"Average Age: {age_sum} / {patient_count} = {average_age:.1f} years")
print(f"Average Heart Rate: {heart_rate_sum} / {patient_count} = {average_heart_rate:.1f} bpm")
```
